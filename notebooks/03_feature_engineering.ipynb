```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
print("Libraries imported successfully")
```

```python
df = pd.read_csv("../data/fraudlens_features.csv")
print("Dataset Loaded successfully")
print("Shape",df.shape)
```

```python
print("Dataset shape",df.shape)
print("Column names:")
print(df.columns.tolist())
```

```python
print(df.dtypes)
```

```python
df["TransactionHour"]=(df["TransactionDT"]//3600)%24
df["TransactionDay"]=df["TransactionDT"]//(3600*24)
print("Time Features Created Successfully")
print(df[["TransactionHour","TransactionDT","TransactionDay"]].head())
```

```python
df["TransactionAmt_log"] = np.log1p(df["TransactionAmt"])
print("Log Transaction Amount created Successfully")
print(df[["TransactionAmt_log","TransactionAmt"]].head())
```

```python
df["MissingCount"] = df.isnull().sum(axis=1)
print("Missing values feeature created")
print(df["MissingCount"].describe())
```

```python
print("P_emaildomain missing:",df["P_emaildomain"].isnull().sum())
print("R_emaildomain missing:",df["R_emaildomain"].isnull().sum())
```

```python
df["EmailDomainMatch"]=(
    df["P_emaildomain"].fillna("missing")==df["R_emaildomain"].fillna("missing")
).astype(int)
print(df["EmailDomainMatch"].value_counts())
```

```python
new_features = [
    "TransactionHour",
    "TransactionDay",
    "TransactionAmt_log",
    "MissingCount",
    "EmailDomainMatch"
]

print(df[new_features].head())
```

```python
missing =  df.isnull().sum()
missning = missing[missing>0].sort_values(ascending=False)
print("Columns Containing Missing Values")
print(missing)
```

```python
print("Fraud Distribution:")
print(df["isFraud"].value_counts())
print("\nFruad percentage")
print(df["isFraud"].value_counts(normalize=True)*100)
```

```python
print(df.describe().T)
```

```python
selected_features = [
    "TransactionDT",
    "TransactionAmt",
    "TransactionAmt_log",
    "TransactionHour",
    "TransactionDay",
    "ProductCD",
    "card1",
    "card2",
    "card3",
    "card4",
    "card5",
    "card6",
    "addr1",
    "addr2",
    "P_emaildomain",
    "R_emaildomain",
    "EmailDomainMatch",
    "MissingCount",
    "isFraud"
]

df_final = df[selected_features].copy()

print("Final dataset shape:", df_final.shape)
```
