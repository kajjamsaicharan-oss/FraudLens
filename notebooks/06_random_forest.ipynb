```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

print("Day 6 libraries loaded successfully")
```

```python
df = pd.read_csv("../data/fraudlens_features.csv")
print("Dataset Loaded Successfully")
print("Shape",df.shape)
```

```python
x = df.drop("isFraud",axis=1)
y = df["isFraud"]
print("Features Shape",x.shape)
print("Target shape",y.shape)
```

```python
x_numeric = x.select_dtypes(include = np.number).copy()
print("Numeric features shape",x_numeric.shape)
print("\nNumeric columns")
print(x_numeric.columns.tolist())
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="median")
x_imputed = imputer.fit_transform(x_numeric)

print(x_imputed.shape)
```

```python
x_imputed = pd.DataFrame(
    x_imputed,
    columns=x_numeric.columns
)

print("DataFrame created successfully")
print("Shape:", x_imputed.shape)
```

```python
print("Prepared feature shape:",x_imputed.shape)
print("Target Shape",y.shape)
print("Missing values",x_imputed.isnull().sum().sum())
```

```python
print("Target Distribution")
print(y.value_counts())
print("\nTarget percentage")
print(y.value_counts(normalize=True)*100)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(
    x_imputed,
    y,
    test_size = 0.20,
    random_state = 42,
    stratify = y
)
print("Training samples:",x_train.shape[0])
print("Testing samples:",x_test.shape[0])
```

```python
print("Training target distributions")
print(y_train.value_counts())
print("\nTesting target distributions")
print(y_test.value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier
print("Random forest imported succesfully")
```

```python
model_fr = RandomForestClassifier(
    n_estimators = 100,
    max_depth = 15,
    class_weight = "balanced",
    random_state = 42,
    n_jobs = 1
)
print("Random forest model created successfully")
```

```python
model_fr.fit(x_train,y_train)
print("Random forest trained successfully")
```

```python
y_pred_rf = model_fr.predict(x_test)
print("Predictions genrated successfully")
```

```python
y_prob_rf = model_fr.predict(x_test)[:,]
print("Fraud probabilities successfully")
```

```python
from sklearn.metrics import precision_score
precision_rf = precision_score(y_test,y_pred_rf)
print("Random forecast prediction:",round(precision_rf),4)
```

```python
from sklearn.metrics import recall_score
recall_rf = recall_score(y_test,y_pred_rf)
print("Random forecast recall:",round(recall_rf,4))
```

```python
from sklearn.metrics import f1_score
f1_rf = f1_score(y_test,y_pred_rf)
print("Random Forest F1-score",round(f1_rf,4))
```

```python
from sklearn.metrics import roc_auc_score

roc_auc_rf = roc_auc_score(y_test, y_prob_rf)

print("Random Forest ROC-AUC:", round(roc_auc_rf, 4))
```

```python
from sklearn.metrics import accuracy_score
accuracy_rf = accuracy_score(y_test,y_pred_rf)
print("Random Forest accuracy"),round(accuracy_rf,4)
```

```python
print("========== Random Forest Results ==========")
print("Accuracy :", round(accuracy_rf, 4))
print("Precision:", round(precision_rf, 4))
print("Recall   :", round(recall_rf, 4))
print("F1-Score :", round(f1_rf, 4))
print("ROC-AUC  :", round(roc_auc_rf, 4))
```

```python
from sklearn.metrics import confusion_matrix

cm_rf = confusion_matrix(y_test, y_pred_rf)

print("Random Forest Confusion Matrix:")
print(cm_rf)
```

```python
from sklearn.metrics import classification_report

print("Random Forest Classification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python
feature_importance = pd.DataFrame({
    "Feature": x_imputed.columns,
    "Importance": model_fr.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print(feature_importance)
```

```python
print("Top 10 Most Important Features:")
print(feature_importance.head(10))
```
