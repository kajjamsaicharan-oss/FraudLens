```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import(
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)
print("Day 5 libraries loaded successfully")
```

```python
df = pd.read_csv("../data/fraudlens_features.csv")
print("Dataset Loaded successfully")
print("Shape",df.shape)
```

```python
x = df.drop("isFraud",axis=1)
y = df["isFraud"]
print("Features shape:",x.shape)
print("Target shape:",y.shape)
```

```python
x_numeric = x.select_dtypes(include=np.number).copy()

print("Numeric features:", x_numeric.shape)
imputer = SimpleImputer(strategy="median")

x_imputed = imputer.fit_transform(x_numeric)

print("Missing values handled successfully")
print("Final feature shape:", x_imputed.shape)
x_imputed = pd.DataFrame(
    x_imputed,
    columns=x_numeric.columns
)
print("Final feature shape:", x_imputed.shape)
print("Total missing values:", x_imputed.isnull().sum().sum())
```

```python
print("Numeric columns:")
print(x.select_dtypes(include=np.number).columns.tolist())
print("\nCategorical columns:")
print(x.select_dtypes(exclude=np.number).columns.tolist())
```

```python
x_numeric = x.select_dtypes(include = np.number).copy()
print("Numerical feature shape:",x_numeric.shape)
```

```python
imputer = SimpleImputer(strategy="median")
x_imputed = imputer.fit_transform(x_numeric)
print("Missing values handled successfully")
print("Final feature shape:",x_imputed.shape)
```

```python
print("Misisng values after imputation:",np.isnan(x_imputed).sum())
```

```python
x_imputed = pd.DataFrame(
    x_imputed,
    columns = x_numeric.columns
)
print(x_imputed.shape)
print(x_imputed.head())
```

```python
print("Fraud distribution:")
print(y.value_counts())
print("\nFraud percentage")
print(y.value_counts(normalize=True)*100)
```

```python
x_train,x_test,y_train,y_test = train_test_split(
    x_imputed,
    y,
    test_size = 0.20,
    random_state = 42,
    stratify = y
)
print("Training samples:",x_train.shape[0])
print("Testing samples:",x_test.shape[0])
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
print("Feature scaling completed")
```

```python
model_lr = LogisticRegression(
    max_iter = 1000,
    class_weight = "balanced",
    random_state = 42
)
model_lr.fit(x_train_scaled,y_train)
print("Logistic regression trained successfully")
```

```python
y_pred = model_lr.predict(x_test_scaled)
y_prob = model_lr.predict_proba(x_test_scaled)[:,1]
print("Predidctions generated successfully")
```

```python
accuracy = accuracy_score(y_test,y_pred)
precision = precision_score(y_test,y_pred)
recall = recall_score(y_test,y_pred)
f1 = f1_score(y_test,y_pred)
roc_auc = roc_auc_score(y_test,y_prob)
print("=======Day 5 Logistic Regression========")
print("Accuracy:",round(accuracy,4))
print("Precision",round(precision,4))
print("recall:",round(recall,4))
print("F1score",round(f1,4))
print("ROC-AUC:",round(roc_auc,4))
```

```python
cm = confusion_matrix(y_test,y_pred)
print("Confusion Matrix:")
print(cm)
```

```python
print("Classification Report")
print(classification_report(y_test,y_pred))
```
