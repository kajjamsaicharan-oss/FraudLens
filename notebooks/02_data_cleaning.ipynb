```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
print("Libraries loaded successfully")
```

```python
transaction_path = "../data/train_transaction.csv"
df = pd.read_csv(transaction_path)
print("Dataset loaded succesfully")
print("Rows:",len(df))
print("columns:",len(df.columns))
```

```python
print("Shape",df.shape)
print("\nFirst 5 rows:")
display(df.head())
```

```python
print("Total columns:",len(df.columns))
print("\ncolumn names")
print(df.columns.tolist())
```

```python
selected_columns = [
    "TransactionID",
    "TransactionDT",
    "TransactionAmt",
    "ProductCD",
    "card1",
    "card2",
    "card3",
    "card4",
    "card5",
    "card6",
    "addr1",
    "addr2",
    "P_emaildomain",
    "R_emaildomain",
    "isFraud"
]

df_clean = df[selected_columns].copy()

print("Selected dataset shape:", df_clean.shape)
```

```python
print(df.columns.tolist())
```

```python
missing =df_clean.isnull().sum()
```

```python
missing
```

```python
missing = missing[missing>0].sort_values(ascending = False)
print("Columns containing missing values")
print(missing)
```

```python
print("Data types:")
print(df_clean.dtypes)
```

```python
print("\nTransaction Amount")
print(df_clean["TransactionAmt"].describe())
```

```python
print("Duplicate Transaction ids:",df_clean["TransactionID"].duplicated().sum())
```

```python
print("Duplicate rows")
print(df_clean.duplicated().sum())
```

```python
print("Minimum Transaction amount:",df_clean["TransactionAmt"].min())
print("Maximum Transaction amount:",df_clean["TransactionAmt"].max())
print("Negative transaction amount is :",(df_clean["TransactionAmt"]<0).sum())

```

```python
df_clean["Hour"]=(df_clean["TransactionDT"]//3600)%24
print(df_clean[["TransactionDT","Hour"]].head())
```

```python
df_clean["Day"]=(df_clean["TransactionDT"]//(3600*24))+1
df_clean["IsWeekend"]=df_clean["Day"]%7>=5
print(df_clean[["TransactionDT","Day","IsWeekend"]].head())
```

```python
df_clean["TransactionAmt"]=pd.to_numeric(
    df_clean["TransactionAmt"],
    errors = "coerce"
)
```

```python
df_clean["LogAmount"]=np.log1p(df_clean["TransactionAmt"])
```

```python
df_clean["AmountLevel"]=pd.cut(
    df_clean["TransactionAmt"],
    bins = [-np.inf,50,100,250,500,np.inf],
    labels = ["low","medium","high","Very high","Extreme"]
)
```

```python
print(df_clean[["TransactionAmt","LogAmount","AmountLevel"]].head())
```

```python
print("Final datset shape:",df_clean.shape)
print("\nFinal columns")
print(df_clean.columns.tolist())
```

```python
output_path = "../data/fraudlens_features.csv"
df_clean.to_csv(output_path,index=False)
print("Dataset saved successfully")
print(output_path)
```
